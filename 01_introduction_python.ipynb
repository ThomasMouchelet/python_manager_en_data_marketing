## 1.4 Exemple Marketing : Calcul de ROI

Imaginons une campagne publicitaire :
- Budget dépensé : 5000€
- Revenus générés : 15000€

**ROI = (Revenus - Coût) / Coût × 100**

```python
# Calcul du ROI d'une campagne
budget = 5000
revenus = 15000

roi = (revenus - budget) / budget * 100
print("Le ROI de la campagne est de", roi, "%")
```

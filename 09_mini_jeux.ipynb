## 9.2 Aléatoire en Python : le module `random`

Avant de coder un jeu, il faut savoir générer du hasard. Le module `random` est livré avec Python — pas besoin d'installation.

| Fonction | Rôle |
|----------|------|
| `random.randint(a, b)` | Un nombre entier entre `a` et `b` (inclus) |
| `random.choice(liste)` | Un élément pris au hasard dans une liste |
| `random.random()` | Un décimal entre 0 et 1 |

````python
# Quelques exemples
print("Nombre entre 1 et 10 :", random.randint(1, 10))
print("Choix aléatoire    :", random.choice(["pierre", "feuille", "ciseaux"]))
print("Décimal aléatoire  :", random.random())
````

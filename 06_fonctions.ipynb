## 6.1 Qu'est-ce qu'une fonction ?

Une fonction est un **bloc de code réutilisable** qui effectue une tâche spécifique.

**Avantages :**
- Éviter la répétition de code
- Organiser le code de manière logique
- Faciliter la maintenance et les tests

````python
# Sans fonction : code répétitif
prix1 = 100
tva1 = prix1 * 0.20
ttc1 = prix1 + tva1
print(f"Prix TTC : {ttc1}€")

prix2 = 250
tva2 = prix2 * 0.20
ttc2 = prix2 + tva2
print(f"Prix TTC : {ttc2}€")

prix3 = 75
tva3 = prix3 * 0.20
ttc3 = prix3 + tva3
print(f"Prix TTC : {ttc3}€")
````

````python
# Avec fonction : code propre et réutilisable
def calculer_ttc(prix_ht):
    """Calcule le prix TTC à partir du prix HT (TVA 20%)"""
    tva = prix_ht * 0.20
    prix_ttc = prix_ht + tva
    return prix_ttc

# Utilisation
print(f"Prix TTC : {calculer_ttc(100)}€")
print(f"Prix TTC : {calculer_ttc(250)}€")
print(f"Prix TTC : {calculer_ttc(75)}€")
````

## 6.4 Fonctions avec plusieurs paramètres

````python
def calculer_reduction(prix, pourcentage):
    """Calcule le prix après réduction"""
    reduction = prix * (pourcentage / 100)
    prix_final = prix - reduction
    return prix_final

# Utilisation
print(calculer_reduction(100, 20))  # 20% de réduction sur 100€
print(calculer_reduction(250, 15))  # 15% de réduction sur 250€
````

````python
def calculer_roi(budget, revenus):
    """Calcule le ROI d'une campagne marketing"""
    if budget == 0:
        return 0
    roi = ((revenus - budget) / budget) * 100
    return round(roi, 2)

# Test
print(f"ROI : {calculer_roi(5000, 15000)}%")
print(f"ROI : {calculer_roi(3000, 2500)}%")  # ROI négatif
````

## 6.5 Paramètres par défaut

````python
# Paramètre avec valeur par défaut
def calculer_ttc(prix_ht, taux_tva=20):
    """Calcule le prix TTC. TVA par défaut : 20%"""
    prix_ttc = prix_ht * (1 + taux_tva / 100)
    return round(prix_ttc, 2)

# Sans spécifier la TVA (utilise 20%)
print(calculer_ttc(100))

# Avec une TVA différente
print(calculer_ttc(100, 5.5))   # TVA réduite
print(calculer_ttc(100, 10))    # TVA intermédiaire
````

````python
def creer_campagne(nom, canal="email", budget=1000, active=True):
    """Crée une campagne marketing avec des valeurs par défaut"""
    campagne = {
        "nom": nom,
        "canal": canal,
        "budget": budget,
        "active": active
    }
    return campagne

# Différentes façons d'appeler la fonction
camp1 = creer_campagne("Black Friday")
camp2 = creer_campagne("Soldes été", canal="sms", budget=500)
camp3 = creer_campagne("Newsletter", active=False)

print(camp1)
print(camp2)
print(camp3)
````

## 6.6 Retourner plusieurs valeurs

````python
def analyser_ventes(ventes):
    """Analyse une liste de ventes et retourne plusieurs statistiques"""
    total = sum(ventes)
    moyenne = total / len(ventes)
    minimum = min(ventes)
    maximum = max(ventes)
    return total, moyenne, minimum, maximum

# Utilisation
ventes_mensuelles = [1500, 2300, 1800, 2500, 1900, 2100]

total, moy, mini, maxi = analyser_ventes(ventes_mensuelles)

print(f"Total : {total}€")
print(f"Moyenne : {moy:.2f}€")
print(f"Min : {mini}€")
print(f"Max : {maxi}€")
````

````python
# Retourner un dictionnaire (plus explicite)
def analyser_ventes_dict(ventes):
    """Analyse une liste de ventes et retourne un dictionnaire"""
    return {
        "total": sum(ventes),
        "moyenne": sum(ventes) / len(ventes),
        "minimum": min(ventes),
        "maximum": max(ventes),
        "nombre": len(ventes)
    }

stats = analyser_ventes_dict(ventes_mensuelles)
print(stats)
print(f"La moyenne est de {stats['moyenne']:.2f}€")
````

## 6.8 Les fonctions Lambda

Fonctions anonymes courtes, utiles pour des opérations simples.

```python
lambda parametres: expression
```

````python
# Fonction classique
def doubler(x):
    return x * 2

# Équivalent en lambda
doubler_lambda = lambda x: x * 2

print(doubler(5))
print(doubler_lambda(5))
````

````python
# Lambda avec plusieurs paramètres
calculer_ttc = lambda prix, tva=20: prix * (1 + tva / 100)

print(calculer_ttc(100))
print(calculer_ttc(100, 5.5))
````

````python
# Utilisation typique : tri personnalisé
clients = [
    {"nom": "Alice", "ca": 5000},
    {"nom": "Bob", "ca": 3000},
    {"nom": "Charlie", "ca": 7000}
]

# Trier par CA décroissant
clients_tries = sorted(clients, key=lambda x: x["ca"], reverse=True)

for client in clients_tries:
    print(f"{client['nom']} : {client['ca']}€")
````

## 6.9 Fonctions utiles : map, filter

````python
# map : appliquer une fonction à chaque élément
prix_ht = [100, 250, 50, 175]

# Convertir tous les prix en TTC
prix_ttc = list(map(lambda x: x * 1.20, prix_ht))
print(prix_ttc)
````

````python
# filter : filtrer les éléments selon une condition
ventes = [150, 80, 200, 45, 300, 90]

# Garder uniquement les ventes > 100€
grosses_ventes = list(filter(lambda x: x > 100, ventes))
print(grosses_ventes)
````

````python
# Combinaison map + filter
emails = ["  ALICE@GMAIL.COM  ", "bob@yahoo", "  charlie@EMAIL.com"]

# Nettoyer les emails
emails_propres = list(map(lambda e: e.strip().lower(), emails))
print(emails_propres)

# Garder uniquement les emails valides (contiennent @gmail)
emails_gmail = list(filter(lambda e: "@gmail" in e, emails_propres))
print(emails_gmail)
````

## 6.10 Exemple Marketing Complet : Système de Scoring

````python
def calculer_score_email(email):
    """Attribue des points selon le type d'email"""
    email = email.lower()
    if "@gmail" in email or "@yahoo" in email or "@hotmail" in email:
        return 5  # Email personnel
    return 20  # Email professionnel

def calculer_score_engagement(pages_vues, temps_passe, a_telecharge):
    """Calcule le score d'engagement"""
    score = 0
    score += min(pages_vues * 2, 20)  # Max 20 points
    score += min(temps_passe // 60, 15)  # 1 point par minute, max 15
    if a_telecharge:
        score += 25
    return score

def calculer_score_entreprise(taille, secteur):
    """Calcule le score selon l'entreprise"""
    score = 0
    
    # Score taille
    if taille >= 500:
        score += 30
    elif taille >= 100:
        score += 20
    elif taille >= 20:
        score += 10
    
    # Score secteur
    secteurs_prioritaires = ["tech", "finance", "retail"]
    if secteur.lower() in secteurs_prioritaires:
        score += 15
    
    return score

def qualifier_lead(score):
    """Détermine la qualification du lead"""
    if score >= 80:
        return "HOT", "À contacter sous 24h"
    elif score >= 50:
        return "WARM", "Nurturing prioritaire"
    else:
        return "COLD", "À surveiller"

def scorer_lead(lead):
    """Fonction principale de scoring"""
    score = 0
    
    # Score email
    score += calculer_score_email(lead["email"])
    
    # Score engagement
    score += calculer_score_engagement(
        lead["pages_vues"],
        lead["temps_passe"],
        lead["a_telecharge"]
    )
    
    # Score entreprise
    score += calculer_score_entreprise(
        lead["taille_entreprise"],
        lead["secteur"]
    )
    
    # Qualification
    qualification, action = qualifier_lead(score)
    
    return {
        "score": score,
        "qualification": qualification,
        "action": action
    }

# Test avec un lead
lead_test = {
    "nom": "Sophie Martin",
    "email": "sophie.martin@entreprise.com",
    "pages_vues": 8,
    "temps_passe": 420,  # secondes
    "a_telecharge": True,
    "taille_entreprise": 250,
    "secteur": "Tech"
}

resultat = scorer_lead(lead_test)

print("=" * 50)
print(f"SCORING : {lead_test['nom']}")
print("=" * 50)
print(f"Score total : {resultat['score']} points")
print(f"Qualification : {resultat['qualification']}")
print(f"Action : {resultat['action']}")
````

````python
# Scorer plusieurs leads
leads = [
    {"nom": "Alice", "email": "alice@gmail.com", "pages_vues": 3, "temps_passe": 120, "a_telecharge": False, "taille_entreprise": 10, "secteur": "Commerce"},
    {"nom": "Bob", "email": "bob@bigcorp.com", "pages_vues": 12, "temps_passe": 600, "a_telecharge": True, "taille_entreprise": 500, "secteur": "Finance"},
    {"nom": "Charlie", "email": "charlie@startup.io", "pages_vues": 6, "temps_passe": 300, "a_telecharge": True, "taille_entreprise": 50, "secteur": "Tech"},
]

print("\nRÉSULTATS DU SCORING")
print("=" * 50)

for lead in leads:
    resultat = scorer_lead(lead)
    print(f"{lead['nom']:15} | Score: {resultat['score']:3} | {resultat['qualification']}")

# Trier par score
leads_scores = [(lead["nom"], scorer_lead(lead)["score"]) for lead in leads]
leads_scores_tries = sorted(leads_scores, key=lambda x: x[1], reverse=True)

print("\nCLASSEMENT")
print("-" * 30)
for i, (nom, score) in enumerate(leads_scores_tries, 1):
    print(f"{i}. {nom} ({score} pts)")
````

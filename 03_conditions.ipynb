## 3.1 Les opérateurs de comparaison

| Opérateur | Signification | Exemple |
|-----------|---------------|----------|
| `==` | Égal à | `x == 5` |
| `!=` | Différent de | `x != 5` |
| `>` | Supérieur à | `x > 5` |
| `<` | Inférieur à | `x < 5` |
| `>=` | Supérieur ou égal | `x >= 5` |
| `<=` | Inférieur ou égal | `x <= 5` |

````python
# Exemples de comparaisons
age = 25

print(age == 25)   # True
print(age != 30)   # True
print(age > 18)    # True
print(age < 20)    # False
print(age >= 25)   # True
````

````python
# Comparaison de chaînes
statut = "premium"

print(statut == "premium")   # True
print(statut == "Premium")   # False (sensible à la casse)
````

## 3.2 La structure if

```python
if condition:
    # Code exécuté si la condition est vraie
```

**Important** : L'indentation (4 espaces) est obligatoire en Python !

````python
# Exemple simple
panier = 150

if panier >= 100:
    print("Livraison gratuite !")
````

````python
# Avec un panier plus petit
panier = 50

if panier >= 100:
    print("Livraison gratuite !")
    
# Rien ne s'affiche car la condition est fausse
````

## 3.3 La structure if/else

```python
if condition:
    # Code si vrai
else:
    # Code si faux
```

````python
panier = 50

if panier >= 100:
    print("Livraison gratuite !")
else:
    print(f"Ajoutez {100 - panier}€ pour la livraison gratuite")
````

````python
# Exemple : vérification de stock
stock = 0

if stock > 0:
    print(f"En stock ({stock} disponibles)")
else:
    print("Rupture de stock")
````

## 3.4 La structure if/elif/else

Pour tester plusieurs conditions :

```python
if condition1:
    # Code si condition1 vraie
elif condition2:
    # Code si condition2 vraie
elif condition3:
    # Code si condition3 vraie
else:
    # Code si aucune condition vraie
```

````python
# Segmentation client par valeur
valeur_client = 2500

if valeur_client >= 5000:
    segment = "VIP"
elif valeur_client >= 2000:
    segment = "Premium"
elif valeur_client >= 500:
    segment = "Standard"
else:
    segment = "Occasionnel"

print(f"Client classé : {segment}")
````

````python
# Système de réduction par palier
montant_panier = 180

if montant_panier >= 200:
    reduction = 20
elif montant_panier >= 150:
    reduction = 15
elif montant_panier >= 100:
    reduction = 10
else:
    reduction = 0

montant_final = montant_panier - reduction
print(f"Panier : {montant_panier}€")
print(f"Réduction : {reduction}€")
print(f"Total à payer : {montant_final}€")
````

## 3.5 Opérateurs logiques

| Opérateur | Signification | Exemple |
|-----------|---------------|----------|
| `and` | ET (les deux vraies) | `x > 0 and x < 10` |
| `or` | OU (au moins une vraie) | `x < 0 or x > 100` |
| `not` | NON (inverse) | `not x > 0` |

````python
# AND : les deux conditions doivent être vraies
age = 25
revenu = 35000

if age >= 18 and revenu >= 30000:
    print("Éligible au crédit")
else:
    print("Non éligible")
````

````python
# OR : au moins une condition doit être vraie
est_etudiant = True
age = 28

if est_etudiant or age < 26:
    print("Réduction jeune applicable")
else:
    print("Tarif normal")
````

````python
# NOT : inverse la condition
est_abonne = False

if not est_abonne:
    print("Découvrez notre offre d'abonnement !")
````

````python
# Combinaison d'opérateurs
age = 30
panier = 200
est_nouveau_client = True

if (age >= 25 and panier >= 150) or est_nouveau_client:
    print("Offre spéciale débloquée !")
````

## 3.6 L'opérateur ternaire

Syntaxe courte pour des conditions simples :

```python
valeur = resultat_si_vrai if condition else resultat_si_faux
```

````python
# Exemple classique
age = 20
statut = "majeur" if age >= 18 else "mineur"
print(statut)
````

````python
# Application marketing
stock = 5
message_stock = "En stock" if stock > 0 else "Rupture de stock"
print(message_stock)
````

## 3.7 Vérifier le contenu d'une chaîne

````python
# L'opérateur 'in' pour vérifier la présence d'un élément
email = "client@gmail.com"

if "@gmail" in email:
    print("Email Gmail détecté")
elif "@yahoo" in email:
    print("Email Yahoo détecté")
else:
    print("Autre fournisseur email")
````

````python
# Vérifier si une chaîne est vide
commentaire = ""

if commentaire:
    print(f"Commentaire reçu : {commentaire}")
else:
    print("Aucun commentaire")
````

## 3.8 Exemple Marketing Complet : Système de Scoring Lead

````python
# Informations du lead
nom = "Sophie Martin"
email = "sophie.martin@entreprise.com"
poste = "Directrice Marketing"
taille_entreprise = 250
a_visite_pricing = True
nombre_pages_vues = 12
source = "LinkedIn"

# Calcul du score
score = 0

# Points selon l'email
if "@gmail" in email or "@yahoo" in email or "@hotmail" in email:
    score += 5  # Email personnel
else:
    score += 20  # Email professionnel

# Points selon le poste
if "directeur" in poste.lower() or "director" in poste.lower():
    score += 30
elif "manager" in poste.lower() or "responsable" in poste.lower():
    score += 20
else:
    score += 10

# Points selon la taille de l'entreprise
if taille_entreprise >= 500:
    score += 25
elif taille_entreprise >= 100:
    score += 15
elif taille_entreprise >= 20:
    score += 10

# Points selon le comportement
if a_visite_pricing:
    score += 20

if nombre_pages_vues >= 10:
    score += 15
elif nombre_pages_vues >= 5:
    score += 10

# Classification du lead
if score >= 80:
    qualification = "HOT - À contacter immédiatement"
elif score >= 50:
    qualification = "WARM - Nurturing prioritaire"
else:
    qualification = "COLD - À surveiller"

# Affichage du résultat
print("=" * 50)
print("SCORING LEAD")
print("=" * 50)
print(f"Nom : {nom}")
print(f"Email : {email}")
print(f"Poste : {poste}")
print(f"Taille entreprise : {taille_entreprise} employés")
print("-" * 50)
print(f"SCORE : {score} points")
print(f"QUALIFICATION : {qualification}")
print("=" * 50)
````

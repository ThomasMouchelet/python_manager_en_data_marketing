## 4.1 La boucle for

La boucle `for` permet de répéter une action pour chaque élément d'une séquence.

```python
for element in sequence:
    # Actions à répéter
```

````python
# Parcourir une liste de clients
clients = ["Alice", "Bob", "Charlie", "Diana"]

for client in clients:
    print(f"Bonjour {client} !")
````

````python
# Parcourir une chaîne de caractères
mot = "DATA"

for lettre in mot:
    print(lettre)
````

````python
# Calculer le total des ventes
ventes = [1500, 2300, 1800, 2100, 1950]

total = 0
for vente in ventes:
    total = total + vente  # ou total += vente

print(f"Total des ventes : {total}€")
````

## 4.2 La fonction range()

Génère une séquence de nombres.

- `range(n)` : de 0 à n-1
- `range(debut, fin)` : de debut à fin-1
- `range(debut, fin, pas)` : avec un intervalle

````python
# range(5) = 0, 1, 2, 3, 4
for i in range(5):
    print(i)
````

````python
# range(1, 6) = 1, 2, 3, 4, 5
for i in range(1, 6):
    print(f"Mois {i}")
````

````python
# range avec un pas de 2
for i in range(0, 10, 2):
    print(i)  # 0, 2, 4, 6, 8
````

````python
# Compte à rebours
for i in range(5, 0, -1):
    print(i)
print("Lancement de la campagne !")
````

## 4.3 enumerate() - Obtenir l'index et la valeur

````python
# Sans enumerate
produits = ["iPhone", "MacBook", "iPad", "AirPods"]

for produit in produits:
    print(produit)
````

````python
# Avec enumerate - on obtient aussi la position
produits = ["iPhone", "MacBook", "iPad", "AirPods"]

for index, produit in enumerate(produits):
    print(f"{index + 1}. {produit}")
````

````python
# Exemple : Classement des meilleurs vendeurs
vendeurs = [
    ("Marie", 45000),
    ("Jean", 38000),
    ("Sophie", 52000),
    ("Pierre", 41000)
]

# Trier par ventes (décroissant)
vendeurs_tries = sorted(vendeurs, key=lambda x: x[1], reverse=True)

print("=== CLASSEMENT DES VENDEURS ===")
for position, (nom, ventes) in enumerate(vendeurs_tries, start=1):
    print(f"{position}. {nom} : {ventes}€")
````

## 4.4 La boucle while

Répète une action **tant que** la condition est vraie.

```python
while condition:
    # Actions à répéter
```

**Attention** : Risque de boucle infinie si la condition ne devient jamais fausse !

````python
# Compteur simple
compteur = 0

while compteur < 5:
    print(f"Compteur : {compteur}")
    compteur += 1  # IMPORTANT : incrémenter pour éviter boucle infinie
````

````python
# Simulation : atteindre un objectif de ventes
objectif = 10000
ventes_actuelles = 0
jour = 1

while ventes_actuelles < objectif:
    vente_jour = 1500  # Ventes quotidiennes
    ventes_actuelles += vente_jour
    print(f"Jour {jour} : {ventes_actuelles}€")
    jour += 1

print(f"\nObjectif atteint en {jour - 1} jours !")
````

## 4.5 break et continue

- `break` : Sort immédiatement de la boucle
- `continue` : Passe à l'itération suivante

````python
# break : arrêter dès qu'on trouve ce qu'on cherche
emails = ["alice@gmail.com", "bob@yahoo.com", "admin@entreprise.com", "charlie@gmail.com"]

for email in emails:
    if "@entreprise" in email:
        print(f"Email professionnel trouvé : {email}")
        break
    print(f"Vérification de {email}...")
````

````python
# continue : ignorer certains éléments
ventes = [1200, -50, 800, 1500, -200, 900]  # Les négatifs sont des retours

total_ventes = 0
for vente in ventes:
    if vente < 0:
        continue  # On ignore les retours pour ce calcul
    total_ventes += vente

print(f"Total des ventes (hors retours) : {total_ventes}€")
````

## 4.6 Boucles imbriquées

````python
# Table de multiplication
for i in range(1, 4):
    for j in range(1, 4):
        print(f"{i} x {j} = {i * j}")
    print("---")
````

````python
# Exemple marketing : Croiser segments et canaux
segments = ["Premium", "Standard", "Occasionnel"]
canaux = ["Email", "SMS", "Push"]

print("Campagnes à créer :")
print("-" * 40)

for segment in segments:
    for canal in canaux:
        print(f"Campagne {canal} pour segment {segment}")
````

## 4.7 Compréhension de liste (List Comprehension)

Syntaxe concise pour créer des listes avec une boucle :

```python
nouvelle_liste = [expression for element in sequence]
```

````python
# Méthode classique
carres = []
for i in range(1, 6):
    carres.append(i ** 2)
print(carres)
````

````python
# Avec list comprehension (plus élégant)
carres = [i ** 2 for i in range(1, 6)]
print(carres)
````

````python
# Avec condition : filtrer les emails Gmail
emails = ["alice@gmail.com", "bob@yahoo.com", "charlie@gmail.com", "diana@outlook.com"]

emails_gmail = [email for email in emails if "@gmail" in email]
print(emails_gmail)
````

````python
# Transformer des données : nettoyer des emails
emails_bruts = ["  ALICE@GMAIL.COM  ", "BOB@yahoo.COM", "  charlie@Gmail.com"]

emails_propres = [email.strip().lower() for email in emails_bruts]
print(emails_propres)
````

````python
# Calculer des prix TTC
prix_ht = [100, 250, 50, 175, 300]
tva = 0.20

prix_ttc = [prix * (1 + tva) for prix in prix_ht]
print(prix_ttc)
````

## 4.8 Exemple Marketing Complet : Analyse de Campagne Email

````python
# Données de campagne email
destinataires = [
    {"email": "alice@gmail.com", "ouvert": True, "clique": True, "achat": 150},
    {"email": "bob@yahoo.com", "ouvert": True, "clique": False, "achat": 0},
    {"email": "charlie@gmail.com", "ouvert": False, "clique": False, "achat": 0},
    {"email": "diana@outlook.com", "ouvert": True, "clique": True, "achat": 89},
    {"email": "emma@gmail.com", "ouvert": True, "clique": True, "achat": 0},
    {"email": "frank@yahoo.com", "ouvert": False, "clique": False, "achat": 0},
    {"email": "grace@gmail.com", "ouvert": True, "clique": False, "achat": 0},
    {"email": "henri@outlook.com", "ouvert": True, "clique": True, "achat": 220},
]

# Calcul des métriques
total_envoyes = len(destinataires)
total_ouverts = 0
total_clics = 0
total_achats = 0
revenus = 0

for dest in destinataires:
    if dest["ouvert"]:
        total_ouverts += 1
    if dest["clique"]:
        total_clics += 1
    if dest["achat"] > 0:
        total_achats += 1
        revenus += dest["achat"]

# Calcul des taux
taux_ouverture = (total_ouverts / total_envoyes) * 100
taux_clic = (total_clics / total_envoyes) * 100
taux_conversion = (total_achats / total_envoyes) * 100

# Affichage du rapport
print("=" * 50)
print("RAPPORT DE CAMPAGNE EMAIL")
print("=" * 50)
print(f"Emails envoyés : {total_envoyes}")
print(f"Emails ouverts : {total_ouverts} ({taux_ouverture:.1f}%)")
print(f"Clics : {total_clics} ({taux_clic:.1f}%)")
print(f"Conversions : {total_achats} ({taux_conversion:.1f}%)")
print("-" * 50)
print(f"Revenus générés : {revenus}€")
print(f"Revenu moyen par conversion : {revenus/total_achats:.2f}€" if total_achats > 0 else "N/A")
print("=" * 50)
````

## 2.1 Qu'est-ce qu'une variable ?

Une variable est comme une **boîte** qui contient une valeur. On lui donne un nom pour pouvoir la retrouver.

```python
nom_variable = valeur
```

````python
# Création de variables
nom_client = "Marie Dupont"
age = 28
budget_marketing = 15000.50
est_client_premium = True

print(nom_client)
print(age)
print(budget_marketing)
print(est_client_premium)
````

## 2.2 Les règles de nommage

**Règles obligatoires :**
- Commence par une lettre ou `_`
- Contient uniquement lettres, chiffres, `_`
- Pas d'espaces ni de caractères spéciaux
- Sensible à la casse (`Age` ≠ `age`)

**Bonnes pratiques :**
- Utiliser des noms explicites
- Utiliser le `snake_case` (mots_séparés_par_underscore)

````python
# Bons noms de variables
nombre_clients = 150
taux_conversion = 3.5
email_marketing = "contact@entreprise.com"

# Mauvais noms (à éviter)
# 2eme_client = "Paul"  # Erreur : commence par un chiffre
# nom-client = "Paul"   # Erreur : contient un tiret
# x = 150               # Pas explicite
````

## 2.3 Les types de données principaux

| Type | Description | Exemple |
|------|-------------|----------|
| `str` | Chaîne de caractères (texte) | `"Bonjour"` |
| `int` | Nombre entier | `42` |
| `float` | Nombre décimal | `3.14` |
| `bool` | Booléen (Vrai/Faux) | `True` ou `False` |

````python
# String (chaîne de caractères)
nom_campagne = "Black Friday 2024"
print(type(nom_campagne))
````

````python
# Integer (entier)
nombre_clics = 1542
print(type(nombre_clics))
````

````python
# Float (décimal)
taux_ouverture = 23.5
print(type(taux_ouverture))
````

````python
# Boolean (booléen)
campagne_active = True
a_depasse_budget = False
print(type(campagne_active))
````

## 2.4 Opérations sur les chaînes de caractères

````python
# Concaténation (assemblage de textes)
prenom = "Marie"
nom = "Dupont"
age = 20

nom_complet = prenom + " " + nom  + str(age)
print(nom_complet)

print(f"{prenom} {nom} {age}")
````

````python
# Répétition
separateur = "Thomas" * 20
print(separateur)
````

````python
# f-strings (formatage moderne - très pratique !)
produit = "iPhone 15"
prix = 999
stock = 45

message = f"Le {produit} coûte {prix}€ et il reste {stock} unités en stock."
print(message)
````

````python
# Méthodes utiles sur les strings
email = "  CONTACT@ENTREPRISE.COM  "

print(email.lower())        # Tout en minuscules
print(email.upper())        # Tout en majuscules
print(email.strip())        # Supprime les espaces au début et à la fin
print(email.strip().lower())  # Combinaison
````

## 2.5 Opérations sur les nombres

````python
# Opérations de base
ventes_janvier = 15000
ventes_fevrier = 18000
ventes_mars = 22000

# Total des ventes
total_ventes = ventes_janvier + ventes_fevrier + ventes_mars
print(f"Total des ventes Q1 : {total_ventes}€")

# Moyenne
moyenne = total_ventes / 3
print(f"Moyenne mensuelle : {moyenne}€")
````

````python
# Arrondi
taux = 23.456789
print(round(taux, 2))  # Arrondi à 2 décimales
````

## 2.6 Conversion de types

````python
# String vers Integer
age_texte = "25"
age_nombre = int(age_texte)
print(age_nombre + 5)  # Maintenant on peut faire des calculs
````

````python
# Integer vers String
nombre_clients = 150
message = "Nous avons " + str(nombre_clients) + " clients"
print(message)
````

````python
# Float vers Integer (attention : tronque la partie décimale)
prix_exact = 99.99
prix_arrondi = int(prix_exact)
print(prix_arrondi)  # Affiche 99
````

## 2.7 Exemple Marketing : Fiche Client

````python
# Informations client
id_client = 12345
nom_client = "Jean Martin"
email_client = "jean.martin@email.com"
age_client = 35
panier_moyen = 127.50
est_abonne_newsletter = True
nombre_achats = 12

# Calcul de la valeur client (CLV simplifiée)
valeur_client = panier_moyen * nombre_achats

# Affichage de la fiche
print("=" * 40)
print("FICHE CLIENT")
print("=" * 40)
print(f"ID: {id_client}")
print(f"Nom: {nom_client}")
print(f"Email: {email_client}")
print(f"Age: {age_client} ans")
print(f"Panier moyen: {panier_moyen}€")
print(f"Nombre d'achats: {nombre_achats}")
print(f"Newsletter: {'Oui' if est_abonne_newsletter else 'Non'}")
print("-" * 40)
print(f"Valeur client totale: {valeur_client}€")
print("=" * 40)
````

## 5.1 Les Listes

Une liste est une collection ordonnée d'éléments modifiable.

```python
ma_liste = [element1, element2, element3]
```

````python
# Création de listes
produits = ["iPhone", "MacBook", "iPad", "AirPods"]
prix = [999, 1299, 799, 199]
notes = [4.5, 4.8, 4.3, 4.7]
mixte = ["Produit A", 100, True, 4.5]  # Types différents possibles

print(produits)
print(prix)
````

## 5.2 Accéder aux éléments (indexation)

Les index commencent à **0** en Python !

````python
produits = ["iPhone", "MacBook", "iPad", "AirPods"]
#              0          1        2         3
#             -4         -3       -2        -1  (index négatifs)

print(produits[0])   # Premier élément : iPhone
print(produits[2])   # Troisième élément : iPad
print(produits[-1])  # Dernier élément : AirPods
print(produits[-2])  # Avant-dernier : iPad
````

````python
# Slicing (découpage)
print(produits[0:2])   # Du début jusqu'à l'index 2 (exclu) : ['iPhone', 'MacBook']
print(produits[1:3])   # De l'index 1 à 3 (exclu) : ['MacBook', 'iPad']
print(produits[:2])    # Les 2 premiers
print(produits[2:])    # À partir du 3ème
print(produits[::2])   # Un sur deux
````

## 5.3 Modifier une liste

````python
# Modifier un élément
produits = ["iPhone", "MacBook", "iPad"]
produits[1] = "MacBook Pro"  # Remplace MacBook
print(produits)
````

````python
# Ajouter des éléments
produits = ["iPhone", "MacBook"]

produits.append("iPad")           # Ajoute à la fin
print(produits)

produits.insert(1, "AirPods")     # Insère à l'index 1
print(produits)

produits.extend(["Watch", "TV"])  # Ajoute plusieurs éléments
print(produits)
````

````python
# Supprimer des éléments
produits = ["iPhone", "MacBook", "iPad", "AirPods"]

produits.remove("iPad")    # Supprime par valeur
print(produits)

dernier = produits.pop()   # Supprime et retourne le dernier
print(f"Supprimé : {dernier}")
print(produits)

del produits[0]            # Supprime par index
print(produits)
````

## 5.4 Méthodes utiles sur les listes

````python
# Longueur
clients = ["Alice", "Bob", "Charlie"]
print(f"Nombre de clients : {len(clients)}")
````

````python
# Tri
ventes = [1500, 2300, 800, 3200, 1100]

ventes.sort()                 # Tri croissant (modifie la liste)
print(ventes)

ventes.sort(reverse=True)     # Tri décroissant
print(ventes)
````

````python
# sorted() - retourne une nouvelle liste sans modifier l'originale
ventes = [1500, 2300, 800, 3200, 1100]
ventes_triees = sorted(ventes)
print(f"Original : {ventes}")
print(f"Trié : {ventes_triees}")
````

````python
# Statistiques de base
ventes = [1500, 2300, 800, 3200, 1100]

print(f"Total : {sum(ventes)}€")
print(f"Minimum : {min(ventes)}€")
print(f"Maximum : {max(ventes)}€")
print(f"Moyenne : {sum(ventes) / len(ventes)}€")
````

````python
# Vérifier la présence d'un élément
produits = ["iPhone", "MacBook", "iPad"]

print("iPhone" in produits)    # True
print("Android" in produits)   # False
````

````python
# Compter les occurrences
commandes = ["iPhone", "MacBook", "iPhone", "iPad", "iPhone", "MacBook"]

print(f"Nombre d'iPhone commandés : {commandes.count('iPhone')}")
````

````python
# Trouver l'index d'un élément
produits = ["iPhone", "MacBook", "iPad"]

position = produits.index("MacBook")
print(f"MacBook est à la position {position}")
````

## 5.5 Les Dictionnaires

Un dictionnaire stocke des paires **clé : valeur**.

```python
mon_dict = {"cle1": valeur1, "cle2": valeur2}
```

````python
# Création d'un dictionnaire
client = {
    "nom": "Marie Dupont",
    "email": "marie@email.com",
    "age": 28,
    "premium": True
}

print(client)
````

````python
# Accéder aux valeurs
print(client["nom"])      # Marie Dupont
print(client["email"])    # marie@email.com
````

````python
# Méthode get() - plus sûre (retourne None si la clé n'existe pas)
print(client.get("nom"))           # Marie Dupont
print(client.get("telephone"))     # None
print(client.get("telephone", "Non renseigné"))  # Valeur par défaut
````

````python
# Modifier ou ajouter des valeurs
client = {"nom": "Marie", "email": "marie@email.com"}

client["email"] = "marie.dupont@email.com"  # Modifier
client["telephone"] = "0612345678"          # Ajouter

print(client)
````

````python
# Supprimer des éléments
client = {"nom": "Marie", "email": "marie@email.com", "age": 28}

del client["age"]              # Supprime la clé "age"
email = client.pop("email")    # Supprime et retourne la valeur

print(client)
print(f"Email supprimé : {email}")
````

## 5.6 Parcourir un dictionnaire

````python
client = {
    "nom": "Marie Dupont",
    "email": "marie@email.com",
    "age": 28
}

# Parcourir les clés
for cle in client:
    print(cle)
````

````python
# Parcourir les valeurs
for valeur in client.values():
    print(valeur)
````

````python
# Parcourir clés et valeurs ensemble
for cle, valeur in client.items():
    print(f"{cle} : {valeur}")
````

## 5.7 Combiner Listes et Dictionnaires

````python
# Liste de dictionnaires (très courant !)
clients = [
    {"nom": "Alice", "email": "alice@email.com", "panier_moyen": 150},
    {"nom": "Bob", "email": "bob@email.com", "panier_moyen": 80},
    {"nom": "Charlie", "email": "charlie@email.com", "panier_moyen": 200}
]

# Afficher tous les clients
for client in clients:
    print(f"{client['nom']} - {client['email']} - Panier: {client['panier_moyen']}€")
````

````python
# Dictionnaire avec des listes comme valeurs
ventes_mensuelles = {
    "janvier": [1500, 1800, 2100],
    "février": [1600, 1900, 2000],
    "mars": [1700, 2200, 2400]
}

# Total par mois
for mois, ventes in ventes_mensuelles.items():
    total = sum(ventes)
    print(f"{mois.capitalize()} : {total}€")
````

````python
# Dictionnaire imbriqué
produit = {
    "nom": "iPhone 15",
    "prix": {
        "ht": 999,
        "ttc": 1198.80
    },
    "specs": {
        "ecran": "6.1 pouces",
        "stockage": [128, 256, 512]
    }
}

print(f"Produit : {produit['nom']}")
print(f"Prix TTC : {produit['prix']['ttc']}€")
print(f"Stockages disponibles : {produit['specs']['stockage']}")
````

## 5.8 Exemple Marketing : Base de données clients

````python
# Base de données clients
base_clients = [
    {"id": 1, "nom": "Alice Martin", "email": "alice@gmail.com", "segment": "Premium", "ca_total": 5200},
    {"id": 2, "nom": "Bob Dupont", "email": "bob@yahoo.com", "segment": "Standard", "ca_total": 1800},
    {"id": 3, "nom": "Charlie Petit", "email": "charlie@gmail.com", "segment": "Premium", "ca_total": 7500},
    {"id": 4, "nom": "Diana Leroy", "email": "diana@outlook.com", "segment": "Occasionnel", "ca_total": 450},
    {"id": 5, "nom": "Emma Bernard", "email": "emma@gmail.com", "segment": "Standard", "ca_total": 2100},
]

# Analyse de la base
print("=" * 50)
print("ANALYSE DE LA BASE CLIENTS")
print("=" * 50)

# Nombre total de clients
print(f"Nombre de clients : {len(base_clients)}")

# CA total
ca_total = sum(client["ca_total"] for client in base_clients)
print(f"CA total : {ca_total}€")

# Répartition par segment
segments = {}
for client in base_clients:
    segment = client["segment"]
    if segment in segments:
        segments[segment] += 1
    else:
        segments[segment] = 1

print("\nRépartition par segment :")
for segment, count in segments.items():
    print(f"  - {segment} : {count} clients")

# Top 3 clients
top_clients = sorted(base_clients, key=lambda x: x["ca_total"], reverse=True)[:3]
print("\nTop 3 clients :")
for i, client in enumerate(top_clients, 1):
    print(f"  {i}. {client['nom']} - {client['ca_total']}€")
````

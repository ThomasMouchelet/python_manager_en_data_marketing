## 8.1 Pourquoi Streamlit ?

Avec ipywidgets, l'interface ne fonctionne que **dans Jupyter**. Pour partager un outil avec un manager ou un client, il faut une **vraie application web**.

**Streamlit** transforme un script Python en site web en quelques lignes :
- Pas besoin de connaître HTML/CSS/JavaScript
- Une commande pour lancer l'app : `streamlit run app.py`
- Très utilisé en data marketing pour des dashboards et démos rapides

## 8.3 Comment ça marche ?

Contrairement à un notebook, Streamlit s'exécute dans un **fichier `.py`** lancé avec :

```bash
streamlit run app.py
```

Cela ouvre automatiquement votre navigateur sur `http://localhost:8501`.

💡 **Astuce notebook** : la commande magique `%%writefile` au début d'une cellule sauvegarde son contenu dans un fichier. Très pratique pour préparer un script Streamlit depuis Jupyter.

## 8.4 Votre première app : Hello World

La cellule ci-dessous **écrit** un fichier `app.py`. Elle ne lance pas l'app — il faut ensuite ouvrir un terminal et taper `streamlit run app.py`.

## 8.5 Les éléments d'affichage

| Fonction | Rôle |
|----------|------|
| `st.title()` | Grand titre |
| `st.header()` | Titre de section |
| `st.subheader()` | Sous-titre |
| `st.write()` | Texte, nombre, dataframe… (universel) |
| `st.markdown()` | Texte avec mise en forme markdown |
| `st.metric()` | Carte avec un chiffre clé |
| `st.dataframe()` | Tableau interactif |

````python
%%writefile app.py
import streamlit as st

st.title("📊 Tableau de bord marketing")
st.header("Performances de la semaine")

st.metric(label="Chiffre d'affaires", value="12 450 €", delta="+8 %")
st.metric(label="Nouveaux clients", value="127", delta="+15")
st.metric(label="Taux de conversion", value="3,2 %", delta="-0,4 %")
````

## 8.6 Les widgets de saisie

Comme avec ipywidgets, mais en plus simple :

````python
%%writefile app.py
import streamlit as st

st.title("Simulateur de ROI")

# Widgets
budget = st.slider("Budget de la campagne (€)", 100, 20000, 5000, step=100)
revenus = st.number_input("Revenus générés (€)", value=15000)
canal = st.selectbox("Canal", ["Email", "SMS", "Display", "Social"])
active = st.checkbox("Campagne active", value=True)

# Calcul
roi = ((revenus - budget) / budget) * 100

st.subheader("Résultat")
st.metric("ROI", f"{roi:.1f} %")

if roi > 100:
    st.success("🚀 Excellente campagne !")
elif roi > 0:
    st.info("👍 Campagne rentable")
else:
    st.error("⚠️ Campagne déficitaire")
````

## 8.7 Mise en page : colonnes et sidebar

- `st.sidebar` : barre latérale (idéal pour les paramètres)
- `st.columns(n)` : découpe la page en colonnes

````python
%%writefile app.py
import streamlit as st

st.title("Dashboard campagne")

# Sidebar pour les paramètres
st.sidebar.header("Paramètres")
budget = st.sidebar.slider("Budget (€)", 100, 20000, 5000)
revenus = st.sidebar.number_input("Revenus (€)", value=15000)

# Calculs
marge = revenus - budget
roi = (marge / budget) * 100 if budget else 0

# 3 colonnes pour les KPI
col1, col2, col3 = st.columns(3)
col1.metric("Budget", f"{budget} €")
col2.metric("Revenus", f"{revenus} €")
col3.metric("ROI", f"{roi:.1f} %")
````

## 8.8 Afficher des données et des graphiques

Streamlit affiche directement les `DataFrame` pandas et les graphiques.

````python
%%writefile app.py
import streamlit as st
import pandas as pd

st.title("Ventes par canal")

donnees = pd.DataFrame({
    "canal": ["Email", "SMS", "Display", "Social"],
    "ventes": [12000, 4500, 8200, 15300]
})

st.subheader("Tableau")
st.dataframe(donnees)

st.subheader("Graphique")
st.bar_chart(donnees, x="canal", y="ventes")
````

## 8.9 Projet complet : Mini-dashboard marketing

On combine tout dans une seule app : sidebar de paramètres, KPI, tableau et graphique.

````python
%%writefile app.py
import streamlit as st
import pandas as pd

st.set_page_config(page_title="Dashboard Marketing", page_icon="📊", layout="wide")

st.title("📊 Dashboard Marketing")
st.caption("Analyse simplifiée des campagnes — démo cours INSEEC")

# Sidebar
st.sidebar.header("Filtres")
tva = st.sidebar.slider("TVA (%)", 0, 25, 20)
objectif_roi = st.sidebar.number_input("Objectif de ROI (%)", value=100)

# Données factices
campagnes = pd.DataFrame({
    "campagne": ["Black Friday", "Soldes été", "Newsletter", "Display Q4"],
    "budget": [5000, 3000, 800, 7000],
    "revenus": [18000, 7500, 1200, 9000],
})
campagnes["marge"] = campagnes["revenus"] - campagnes["budget"]
campagnes["roi"] = (campagnes["marge"] / campagnes["budget"]) * 100
campagnes["objectif_atteint"] = campagnes["roi"] >= objectif_roi

# KPI globaux
col1, col2, col3, col4 = st.columns(4)
col1.metric("Budget total", f"{campagnes['budget'].sum():,} €")
col2.metric("Revenus totaux", f"{campagnes['revenus'].sum():,} €")
col3.metric("ROI moyen", f"{campagnes['roi'].mean():.1f} %")
col4.metric("Campagnes OK", f"{campagnes['objectif_atteint'].sum()} / {len(campagnes)}")

st.divider()

# Détail
col_a, col_b = st.columns([1, 1])
with col_a:
    st.subheader("Détail des campagnes")
    st.dataframe(campagnes, use_container_width=True)
with col_b:
    st.subheader("ROI par campagne")
    st.bar_chart(campagnes, x="campagne", y="roi")

# Alerte
perdantes = campagnes[campagnes["roi"] < 0]
if len(perdantes) > 0:
    st.warning(f"⚠️ {len(perdantes)} campagne(s) en perte : {', '.join(perdantes['campagne'])}")
else:
    st.success("✅ Toutes les campagnes sont rentables")
````
